# OpenAI API prompt examples

## Minimal prompt execution

```python
from openai import OpenAI
import getpass

OPENAI_API_KEY = getpass.getpass('Enter your OPENAI_API_KEY') 
```

```python
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
prompt_input = """โปรดระวังการโจมตีแบบฟิชชิง (Phishing) และตรวจสอบข้อมูลอย่างละเอียดทุกครั้งก่อนคลิกลิงก์หรือกรอกข้อมูลส่วนตัว"""
response = client.chat.completions.create(
  model="gpt-5-nano",
  messages=[
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": prompt_input}
  ] 
)

print(response)
```

```python
print(response.choices[0].message.content)
```

# Running prompts with LangChain

## Basic prompt with LangChain

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(openai_api_key=OPENAI_API_KEY,
                 model_name="gpt-5-nano")
```

# Prompt templates

## Prompt template - implementing it with a Python function

```python
def generate_text_summary_prompt(text, num_words, tone):
    return f"คุณเป็นก๊อปปี้ไรเตอร์ (Copywriter) ผู้เชี่ยวชาญ จงเขียนสรุปข้อความต่อไปนี้ให้มีความยาว {num_words} คำ โดยใช้โทนเสียงแบบ {tone}: {text}"
```

```python
segovia_aqueduct_text = """สะพานส่งน้ำแห่งเซโกเบีย (สเปน: Acueducto de Segovia) เป็นสะพานส่งน้ำโรมันโบราณที่ตั้งอยู่ในเมืองเซโกเบีย ประเทศสเปน ถูกสร้างขึ้นราวคริสต์ศตวรรษที่ 1 เพื่อส่งน้ำจากแหล่งน้ำพุในภูเขาที่อยู่ห่างออกไป 17 กิโลเมตร (11 ไมล์) มายังน้ำพุ โรงอาบน้ำสาธารณะ และบ้านเรือนส่วนบุคคลในเมือง และมีการใช้งานมาอย่างยาวนานจนถึงปี ค.ศ. 1973

โครงสร้างส่วนที่ยกระดับสูงขึ้นมา ซึ่งประกอบด้วยแนวซุ้มโค้งที่สมบูรณ์จำนวน 167 ส่วน ถือเป็นหนึ่งในสะพานส่งน้ำโรมันที่ได้รับการอนุรักษ์ไว้เป็นอย่างดีที่สุด และเป็นสัญลักษณ์ที่สำคัญที่สุดของเมืองเซโกเบีย ดังจะเห็นได้จากการปรากฏอยู่บนตราอาร์มประจำเมือง เมืองเก่าเซโกเบียและสะพานส่งน้ำแห่งนี้ได้รับการขึ้นทะเบียนเป็นแหล่งมรดกโลกโดยองค์การยูเนสโกในปี ค.ศ. 1985 เนื่องจากสะพานส่งน้ำนี้ไม่มีจารึกที่อ่านออกได้อย่างชัดเจน (ซึ่งดูเหมือนว่าเคยมีจารึกอยู่บริเวณส่วนบนสุดของโครงสร้าง) ทำให้ไม่สามารถระบุวันที่ก่อสร้างได้อย่างแน่ชัด ช่วงเวลาในการก่อสร้างสะพานส่งน้ำแห่งนี้เคยเป็นปริศนามาอย่างยาวนาน แม้ว่าจะเคยเชื่อกันว่าสร้างขึ้นในช่วงคริสต์ศตวรรษที่ 1 ในรัชสมัยของจักรพรรดิโดมิเชียน, จักรพรรดิเนอร์วา และจักรพรรดิตรายัน จนกระทั่งในปลายคริสต์ศตวรรษที่ 20 เกลซา อัลเฟิลดี (Géza Alföldy) ได้ถอดรหัสข้อความบนแผ่นจารึกอุทิศ โดยการศึกษาจากรอยยึดที่เคยใช้ตรึงตัวอักษรสำริดซึ่งในปัจจุบันได้สูญหายไปแล้ว เขาได้ข้อสรุปว่าจักรพรรดิโดมิเชียน (ค.ศ. 81–96) เป็นผู้มีพระบรมราชโองการให้สร้างขึ้น และมีการเสนอว่าปี ค.ศ. 98 น่าจะเป็นปีที่สร้างเสร็จสมบูรณ์มากที่สุด อย่างไรก็ตาม ในปี ค.ศ. 2016 ได้มีการเผยแพร่หลักฐานทางโบราณคดีที่บ่งชี้ถึงช่วงเวลาที่หมวดหมู่หลังจากนั้นเล็กน้อย คือหลังปี ค.ศ. 112 เป็นต้นไป ในช่วงรัฐบาลของจักรพรรดิตรายัน หรือในช่วงต้นรัชสมัยของจักรพรรดิฮาดริอานุส ตั้งแต่ปี ค.ศ. 117"""

input_prompt = generate_text_summary_prompt(
    text=segovia_aqueduct_text, 
    num_words=20,
    tone="รอบรู้และน่าสนใจ")

response = llm.invoke(input_prompt)
print(response.content)
```

## Prompt template - using LangChain's ChatPromptTemplate

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template(
"""คุณเป็นก๊อปปี้ไรเตอร์ (Copywriter) ผู้เชี่ยวชาญ จงเขียนสรุปข้อความต่อไปนี้ให้มีความยาว {num_words} คำ โดยใช้โทนเสียงแบบ {tone}: {text}""")
```

```python
prompt = prompt_template.format(
    text=segovia_aqueduct_text,
    num_words=20, 
    tone="รอบรู้และน่าสนใจ")
```

```python
response = llm.invoke(prompt)
print(response.content)
```

# Few shot prompt with LangChain

## Using standard prompt

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(openai_api_key=OPENAI_API_KEY,
                 model_name="gpt-5-mini")
```

```python
prompt_input = """จัดหมวดหมู่ตัวเลขต่อไปนี้เป็น Abra, Kadabra หรือ Abra Kadabra:
3, 4, 5, 7, 8, 10, 11, 13, 35
ตัวอย่าง:
6 // หารด้วย 5 ไม่ลงตัว, หารด้วย 7 ไม่ลงตัว // ไม่มี
15 // หารด้วย 5 ลงตัว, หารด้วย 7 ไม่ลงตัว // Abra
12 // หารด้วย 5 ไม่ลงตัว, หารด้วย 7 ไม่ลงตัว // ไม่มี
21 // หารด้วย 5 ไม่ลงตัว, หารด้วย 7 ลงตัว // Kadabra
70 // หารด้วย 5 ลงตัว, หารด้วย 7 ลงตัว // Abra Kadabra
"""

response = llm.invoke(prompt_input)
print(response.content)
```

## Using Langchain's FewShotPromptTemplate

```python
from langchain_core.prompts.few_shot import FewShotPromptTemplate
from langchain_core.prompts.prompt import PromptTemplate

examples = [
  {
      "number": 6,
      "reasoning": "หารด้วย 5 และ 7 ไม่ลงตัว",
      "result": "None"
  },
  {
      "number": 15,
      "reasoning": "หารด้วย 5 ลงตัว แต่หารด้วย 7 ไม่ลงตัว",
      "result": "Abra"
  },
  {
      "number": 12,
      "reasoning": "หารด้วย 5 และ 7 ไม่ลงตัว",
      "result": "None"
  },
  {
      "number": 21,
      "reasoning": "หารด้วย 7 ลงตัว แต่หารด้วย 5 ไม่ลงตัว",
      "result": "Kadabra"
  },
  {
      "number": 70,
      "reasoning": "หารด้วย 5 และ 7 ลงตัว",
      "result": "Abra Kadabra"
  } ]

example_prompt = PromptTemplate(input_variables=["number", "reasoning", "result"], template="{number} \\ {reasoning} \\ {result}")
few_shot_prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="จัดหมวดหมู่ตัวเลขต่อไปนี้เป็น Abra, Kadabra หรือ Abra Kadabra: {comma_delimited_input_numbers}",
    input_variables=["comma_delimited_input_numbers"]
)

prompt_input = few_shot_prompt.format(comma_delimited_input_numbers="3, 4, 5, 7, 8, 10, 11, 13, 35.")

response = llm.invoke(prompt_input)
print(response.content)
```
